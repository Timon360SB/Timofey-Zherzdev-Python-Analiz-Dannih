## 1.2

Создайте DataFrame из словаря

```python
import pandas as pd
import numpy as np

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR'],
    'salary': [80000, 60000, 90000, 75000, np.nan, 85000, 70000, 62000],
    'bonus': [5000, 3000, np.nan, 4000, 2000, 6000, 3500, np.nan],
    'hire_date': pd.to_datetime(['2020-01-15', '2019-03-22', '2021-07-01', '2018-11-05',
                                 '2022-02-10', '2020-09-30', '2019-12-12', '2021-05-20'])
}

df = pd.DataFrame(data)
print(df)
```

## 1.3

Выведите первые и последние строки, информацию о типах данных, статистику

```python
print("Первые строки:")
print(df.head())
print("Последние строки:")
print(df.tail())
print("Информация о таблице:")
df.info()
print("Статистика:")
print(df.describe())
```

## 2.4

Создайте новый столбец total_comp = salary + bonus (пропуски в bonus считать нулём)

```python
df["total_comp"] = df["salary"] + df["bonus"].fillna(0)
print(df)
```

## 2.5

Удалите столбец hire_date

```python
df = df.drop(columns="hire_date")
print(df)
```

## 3.1

Подсчитайте количество пропусков в каждом столбце

```python
print(df.isna().sum())
```

## 3.2

Заполните пропуски в salary средним значением по отделу

```python
average_salary = df.groupby("department")["salary"].transform("mean")
df["salary"] = df["salary"].fillna(average_salary)
print(df)
```

## 3.3

Заполните пропуски в bonus нулём

```python
df["bonus"] = df["bonus"].fillna(0)

# Пересчитываем компенсацию после заполнения пропусков.
df["total_comp"] = df["salary"] + df["bonus"]
print(df)
```

## 3.4

Удалите строки, гдеsalary всё ещё пропущен (если остались)

```python
df = df.dropna(subset=["salary"])
print(df)
```

## 4.1

Найдите среднюю зарплату по каждому отделу

```python
print(df.groupby("department")["salary"].mean())
```

## 4.2

Для каждого отдела вычислите: среднюю зарплату, максимальный бонус, количество сотрудников

```python
print(df.groupby("department").agg({
    "salary": "mean",
    "bonus": "max",
    "name": "count"
}))
```

## 4.3

Создайте сводную таблицу: индекс - отдел, значения - средняя зарплата

```python
print(df.pivot_table(index="department", values="salary", aggfunc="mean"))
```

## 4.4

Отсортируйте отделы по средней зарплате по убыванию

```python
average_salary = df.groupby("department")["salary"].mean()
print(average_salary.sort_values(ascending=False))
```

## 5.1

Создайте DataFrame managers с отделами и руководителями

```python
import pandas as pd

managers = pd.DataFrame({
    "department": ["IT", "HR", "Finance"],
    "manager": ["Иванов", "Петрова", "Сидоров"]
})
print(managers)
```

## 5.2

Объедините df с managers по столбцу department (left join)

```python
import pandas as pd

df = pd.merge(df, managers, on="department", how="left")
print(df)
```

## 5.3

Создайте новую строку с данными о сотруднике и добавьте её к df через concat

```python
import pandas as pd

new_employee = pd.DataFrame({
    "name": ["Ivan"],
    "department": ["IT"],
    "salary": [78000],
    "bonus": [4000],
    "manager": ["Иванов"]
})
new_employee["total_comp"] = new_employee["salary"] + new_employee["bonus"]

df = pd.concat([df, new_employee])
print(df)
```

## 5.4

Сбросьте индекс после объединения

```python
df = df.reset_index(drop=True)
print(df)
```

## 6.1

Найдите топ-3 сотрудников по общей компенсации (total_comp)
