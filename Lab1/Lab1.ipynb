## 1.1

Создайте массив float32 со значениями [0.1, 0.2, 0.3] и вычислите сумму. Сравните с точным значением

```python
import numpy as np

a = np.array([0.1, 0.2, 0.3], dtype=np.float32)
s = a.sum()

print("Сумма:", float(s))
print("Точное значение: 0.6")
print("Ошибка:", abs(float(s) - 0.6))
```

## 3.1

Создайте на диске memory-mapped файл для массива формы (1000, 1000) типа float64. Заполните его случайными числами

```python
m = np.memmap("matrix.dat", dtype="float64", mode="w+", shape=(1000, 1000))
rng = np.random.default_rng(1520441)

for i in range(1000):
    m[i] = rng.random(1000)

m.flush()
print("Форма:", m.shape)
print("Размер:", m.nbytes, "байт")
```

## 3.2

Прочитайте отдельный столбец и строку из этого файла (без загрузки всего массива)

```python
m = np.memmap("matrix.dat", dtype="float64", mode="r", shape=(1000, 1000))
row = m[0, :]
column = m[:, 0]

print("Строка:", row[:5])
print("Столбец:", column[:5])
```

## 3.3

Сравните время обхода массива по строкам (row-major) и по столбцам (column-major) с использованием %timeit. Объясните результат.

```python
def sum_rows(a):
    s = 0.0
    for i in range(a.shape[0]):
        for j in range(a.shape[1]):
            s += a[i, j]
    return s


def sum_columns(a):
    s = 0.0
    for j in range(a.shape[1]):
        for i in range(a.shape[0]):
            s += a[i, j]
    return s


a = np.asarray(m)
print("По строкам:")
%timeit -n 1 -r 3 sum_rows(a)
print("По столбцам:")
%timeit -n 1 -r 3 sum_columns(a)
```

## 4.1

Создайте разреженную матрицу размера (5, 5) с плотностью ~20% в формате CSR

```python
from scipy import sparse

S = sparse.random(5, 5, density=0.2, format="csr", random_state=42)
print(S.toarray())
print("Ненулевых элементов:", S.nnz)
```

## 4.2

Преобразуйте её в COO и CSC

```python
S_coo = S.tocoo()
S_csc = S.tocsc()

print("COO:")
print(S_coo)
print("CSC:")
print(S_csc)
```

## 4.3

Выполните умножение разреженной матрицы на плотный вектор

```python
x = np.arange(1, 6)
print("Вектор:", x)
print("Результат:", S @ x)
```

## 4.4

Сравните время умножения разреженной и плотной матриц при большой размерности (например, 1000×1000 с плотностью 0.01)

```python
S_big = sparse.random(1000, 1000, density=0.01, format="csr", random_state=42)
A_big = S_big.toarray()
x_big = np.ones(1000)

print("CSR:")
%timeit -n 100 -r 3 S_big @ x_big
print("Плотная матрица:")
%timeit -n 100 -r 3 A_big @ x_big
```

## 5.1

Напишите функцию на Python, которая вычисляет сумму квадратов элементов списка

```python
def sum_squares_python(values):
    s = 0.0
    for x in values:
        s += x * x
    return s


print(sum_squares_python([1.0, 2.0, 3.0]))
```

## 5.2

Напишите аналогичную функцию на Cython (в отдельном файле .pyx) и скомпилируйте её

```python
%load_ext Cython
```

```python
%%cython

def sum_squares_cython(list values):
    cdef double s = 0.0
    cdef double x

    for x in values:
        s += x * x

    return s
```

```python
print(sum_squares_cython([1.0, 2.0, 3.0]))
```

## 5.3

Сравните время выполнения обеих функций на массиве из 10⁷ элементов

```python
from time import perf_counter

values = np.linspace(0, 1, 10**7).tolist()

start = perf_counter()
result_python = sum_squares_python(values)
time_python = perf_counter() - start

start = perf_counter()
result_cython = sum_squares_cython(values)
time_cython = perf_counter() - start

print("Результат Python:", result_python)
print("Результат Cython:", result_cython)
print("Время Python:", time_python, "с")
print("Время Cython:", time_cython, "с")
print("Ускорение:", time_python / time_cython)
```
